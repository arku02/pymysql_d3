```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By


options = Options()
options.add_argument("--start-maximized") # Chrome 瀏覽器在啟動時最大化視窗
options.add_argument("--incognito") # 無痕模式
options.add_argument("--disable-popup-blocking") # 停用 Chrome 的彈窗阻擋功能


# 建立 Chrome 的 webdriver 物件
driver = webdriver.Chrome(options=options)
product = "iphone 17"

url = f"https://24h.pchome.com.tw/search/?q={product}"
# 開啟網頁
driver.get(url)
```

```python
driver.quit()
```

```python

elements = driver.find_elements(By.CSS_SELECTOR,'ul.c-listInfoGridlist.c-listInfoGridlist--wrapProdCard.is-cardGap16X12')

for element in elements:
    img_element = element.find_element(By.CSS_SELECTOR, 'div.c-prodInfoV2img > img')

    img_url = img_element.get_attribute('src')

    title = img_element.get_attribute('alt')

    price_element = element.find_element(By.CSS_SELECTOR, 'div.c-prodInfoV2priceValue ').text.replace('$', '').replace(',', '')

    if price_element == []:
        continue


    price = element.find_element(By.CSS_SELECTOR, 'div.c-prodInfoV2priceValue ').text.replace('$', '').replace(',', '')

    product_link = element.find_element(By.CSS_SELECTOR, 'a.c-prodInfoV2link.gtmClickV2').get_attribute('href')



    print(img_element.get_attribute('outerHTML'))
    print(img_url)
    print(title)
    print(price)
    print(product_link)

    break

next_page_btn = driver.find_element(By.CSS_SELECTOR, 'div.c-pagination__button.is-next > button')
print(next_page_btn.get_attribute('outerHTML'))

# 可以按 > none, 到最後一頁 -> True
if next_page_btn.get_attribute('disabled'):
    print('\033[31m')
next_page_btn.click()
```
