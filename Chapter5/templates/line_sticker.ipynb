```python
import requests
from bs4 import BeautifulSoup as bs
url = "https://store.line.me/stickershop/product/33904260/zh-Hant"

res = requests.get(url)

print(res)

soup = bs(res.text, "lxml")
```

```python
import json
import os

folder = 'line_stickers'
if not os.path.exists(folder):
    os.mkdir(folder)
elements = soup.select("ul.mdCMN09Ul.FnStickerList > li")
print(len(elements))

n = 0
for element in elements:
    print(element)
    img_element = json.loads(element['data-preview'])
    img_url = img_element['staticUrl']
    print(img_url)
    n += 1
    with open(f'{folder}/sticker_{n}.png', mode='wb') as img:
        img.write(requests.get(img_url).content)
    
    
```
